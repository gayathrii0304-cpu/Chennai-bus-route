```python
import numpy as np
```

```python
# Convert latitude and longitude to radians

lat = np.radians(stop_data["Lat"])
lng = np.radians(stop_data["Lng"])

# Calculate distance from the previous stop

dlat = lat.diff()
dlng = lng.diff()

a = (
    np.sin(dlat / 2) ** 2
    + np.cos(lat) * np.cos(lat.shift(1)) * np.sin(dlng / 2) ** 2
)

distance_km = 6371 * 2 * np.arcsin(np.sqrt(a))

stop_data["Distance_From_Previous_km"] = distance_km

print(stop_data.head(10))
```

```python
print("Minimum distance:",
      stop_data["Distance_From_Previous_km"].min())

print("Maximum distance:",
      stop_data["Distance_From_Previous_km"].max())

print("Average distance:",
      stop_data["Distance_From_Previous_km"].mean())
```

```python
large_distances = stop_data[
    stop_data["Distance_From_Previous_km"] > 10
]

print(large_distances[
    ["Stop Name", "Lat", "Lng", "Distance_From_Previous_km"]
].head(20))
```

```python
plt.figure(figsize=(10, 5))

plt.hist(
    stop_data["Distance_From_Previous_km"].dropna(),
    bins=30
)

plt.title("Distribution of Distance Between Stops")
plt.xlabel("Distance (km)")
plt.ylabel("Number of Stops")

plt.tight_layout()
plt.show()
```

```python
def distance_category(distance):
    if pd.isna(distance):
        return "Unknown"
    elif distance < 1:
        return "Short"
    elif distance < 3:
        return "Medium"
    else:
        return "Long"

stop_data["Distance_Category"] = (
    stop_data["Distance_From_Previous_km"]
    .apply(distance_category)
)

print(stop_data[
    ["Stop Name", "Distance_From_Previous_km", "Distance_Category"]
].head(20))
```

```python
delay_data = pd.read_csv("chennai_bus_delay_data.csv")

print(delay_data.head())
print(delay_data.shape)
```

```python
delay_data["Date"] = pd.to_datetime(
    delay_data["Date"],
    errors="coerce"
)

delay_data["Scheduled_Time"] = pd.to_datetime(
    delay_data["Scheduled_Time"],
    format="%H:%M",
    errors="coerce"
)

delay_data["Actual_Time"] = pd.to_datetime(
    delay_data["Actual_Time"],
    format="%H:%M",
    errors="coerce"
)

print(delay_data.dtypes)
```

```python
delay_data["Delay_Minutes"] = (
    delay_data["Actual_Time"] -
    delay_data["Scheduled_Time"]
).dt.total_seconds() / 60

print(
    delay_data[
        ["Route_Id", "Stop_Id",
         "Scheduled_Time", "Actual_Time",
         "Delay_Minutes"]
    ].head(10)
)
```

```python
delay_data["Hour"] = (
    delay_data["Scheduled_Time"].dt.hour
)

delay_data["Peak_Hour"] = np.where(
    delay_data["Hour"].isin([7, 8, 9, 17, 18, 19]),
    "Yes",
    "No"
)

print(
    delay_data[
        ["Scheduled_Time", "Hour", "Peak_Hour"]
    ].head(10)
)
```

```python
delay_data["Day_of_Week"] = (
    delay_data["Date"].dt.day_name()
)

print(
    delay_data[
        ["Date", "Day_of_Week"]
    ].head(10)
)
```

```python
print("Number of delay records:", len(delay_data))

print("\nColumns:")
print(delay_data.columns.tolist())

print("\nFirst 10 records:")
print(delay_data.head(10))
```

```python
print("Average Delay:",
      round(delay_data["Delay_Minutes"].mean(), 2), "minutes")

print("Maximum Delay:",
      delay_data["Delay_Minutes"].max(), "minutes")

print("Minimum Delay:",
      delay_data["Delay_Minutes"].min(), "minutes")
```

```python
def delay_category(minutes):
    if minutes <= 0:
        return "On Time"
    elif minutes <= 5:
        return "Slight Delay"
    elif minutes <= 15:
        return "Moderate Delay"
    else:
        return "High Delay"

delay_data["Delay_Category"] = (
    delay_data["Delay_Minutes"]
    .apply(delay_category)
)

print(
    delay_data["Delay_Category"].value_counts()
)
```

```python
delay_counts = delay_data["Delay_Category"].value_counts()

print("Delay Category Counts:")
print(delay_counts)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    delay_counts.index.astype(str),
    delay_counts.values
)

plt.title("Chennai Bus Delay Categories")
plt.xlabel("Delay Category")
plt.ylabel("Number of Trips")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
